```python
import os
import nbformat

from nbclient import NotebookClient
from datetime import datetime
```

```python
PROJECT_PATH = "/content/drive/MyDrive/Colab_Notebooks/Desafio_Fifa_World_Cup"
print(PROJECT_PATH)
```

```python
notebooks = [
    "Desafio_Fifa_World_Cup_Origem_1.ipynb"
    "Desafio_Fifa_World_Cup_Bronze_2.ipynb",
    "Desafio_Fifa_World_Cup_Silver_3.ipynb",
    "Desafio_Fifa_World_Cup_Gold_4.ipynb",
    "04_Gold_Modelagem.ipynb"
]
```

```python
def executar_notebook(nome_notebook):

    caminho = os.path.join(PROJECT_PATH, nome_notebook)

    print("=" * 70)
    print(f"🚀 EXECUTANDO: {nome_notebook}")
    print(f"🕐 Início: {datetime.now()}")
    print("=" * 70)

    if not os.path.exists(caminho):
        raise FileNotFoundError(
            f"Notebook não encontrado: {caminho}"
        )

    with open(caminho, "r", encoding="utf-8") as f:
        notebook = nbformat.read(f, as_version=4)

    cliente = NotebookClient(
        notebook,
        timeout=3600,
        kernel_name="python3"
    )

    cliente.execute()

    print("=" * 70)
    print(f"✅ CONCLUÍDO: {nome_notebook}")
    print(f"🕐 Fim: {datetime.now()}")
    print("=" * 70)
```

```python
notebooks = [
    "Desafio_Fifa_World_Cup_Origem_1.ipynb",
    "Desafio_Fifa_World_Cup_Bronze_2.ipynb",
    "Desafio_Fifa_World_Cup_Silver_3.ipynb",
    "Desafio_Fifa_World_Cup_Gold_4.ipynb"
]
```

```python
inicio = datetime.now()

print("\n")
print("🏆 DESAFIO FIFA WORLD CUP")
print("🚀 INICIANDO PIPELINE")
print(f"🕐 {inicio}")
print("\n")

for notebook in notebooks:

    try:

        executar_notebook(notebook)

    except Exception as erro:

        print("\n❌ ERRO NO PIPELINE")
        print(f"Notebook: {notebook}")
        print(f"Erro: {erro}")

        raise

fim = datetime.now()

print("\n")
print("=" * 70)
print("🏆 PIPELINE FIFA WORLD CUP FINALIZADO COM SUCESSO")
print("=" * 70)
print(f"🕐 Início:  {inicio}")
print(f"🕐 Fim:     {fim}")
print(f"⏱️ Duração: {fim - inicio}")
print("=" * 70)
```
